# Multiplicative Neurons

A Neural Zoo notebook covering intuition, mathematics, implementations, visualization ideas, and a mini project.

# 1. First Principles

Traditional neurons **add** weighted inputs:

\[
z=\sum_i w_i x_i+b
\]

Multiplicative neurons instead model **feature interactions** by multiplying transformed inputs:

\[
y=\prod_i (w_i x_i+b_i)
\]

Addition accumulates evidence. Multiplication captures interactions between features.


# 2. Mathematical Formulation

Traditional:

\[
z=\sum_i w_i x_i+b
\]

Multiplicative:

\[
y=\prod_i (w_i x_i+b_i)
\]

Log-space interpretation:

\[
\log y=\sum_i \log(w_i x_i+b_i)
\]

This reveals a connection between multiplication and addition.


```python
import numpy as np

x=np.array([2.,3.,1.])
w=np.array([0.5,0.2,1.5])
b=np.array([0.1,0.1,0.1])

y=np.prod(w*x+b)
print("Multiplicative output:",y)
```

# 3. NumPy Implementation

```python
class MultiplicativeNeuron:

    def __init__(self,input_size):
        self.w=np.random.randn(input_size)
        self.b=np.zeros(input_size)

    def forward(self,x):
        return np.prod(self.w*x+self.b)

m=MultiplicativeNeuron(4)
print(m.forward(np.random.randn(4)))
```

# 4. PyTorch Implementation

```python
import torch
import torch.nn as nn

class MultiplicativeLayer(nn.Module):

    def __init__(self,in_features,out_features):
        super().__init__()
        self.weight=nn.Parameter(torch.randn(out_features,in_features))
        self.bias=nn.Parameter(torch.zeros(out_features,in_features))

    def forward(self,x):
        x=x.unsqueeze(1)
        y=x*self.weight+self.bias
        return torch.prod(y,dim=2)

layer=MultiplicativeLayer(4,3)
print(layer(torch.randn(2,4)))
```

# 5. Multiplicative MLP

```python
class MultiplicativeMLP(nn.Module):

    def __init__(self):
        super().__init__()
        self.flatten=nn.Flatten()
        self.m1=MultiplicativeLayer(784,128)
        self.m2=MultiplicativeLayer(128,64)
        self.out=nn.Linear(64,10)

    def forward(self,x):
        x=self.flatten(x)
        x=self.m1(x)
        x=torch.relu(x)
        x=self.m2(x)
        x=torch.relu(x)
        return self.out(x)

model=MultiplicativeMLP()
print(model)
```

# 7. Advantages

- Models feature interactions naturally.
- Higher expressive power.
- Useful for gating mechanisms.
- Related to multiplicative RNNs and GLUs.

# 8. Limitations

- Numerical instability.
- Products can explode or vanish.
- Harder to optimize.
- More sensitive to initialization.
