# Roadmap

```
ReLU
   ↓
LeakyReLU
   ↓
PReLU
   ↓
Dynamic ReLU
   ↓
HyperNetworks
```

Dynamic ReLU is the first activation function that **changes depending on the input**.

Instead of having one fixed activation for every image,

the network **creates a new activation function for every input**.

# Why isn't ReLU enough?

ReLU is

\[
f(x)=\max(0,x)
\]

Advantages

- Fast
- Simple
- Sparse activations
- Easy optimization

Problems

- Dead neurons
- Same activation for every input
- Cannot adapt to different features

Suppose the network receives

Image A → Cat

Image B → Airplane

Should both images use exactly the same activation function?

Probably not.

This motivates Dynamic ReLU.

# First Principles

A neuron consists of two parts.

```
Aggregation

↓

Activation
```

Traditional deep learning only improves the

Aggregation

for example

- Attention
- Residual Connections
- Better Convolutions

But the activation function stays almost fixed.

Dynamic ReLU asks

> Why not make the activation itself intelligent?

Instead of

```
Input

↓

ReLU

↓

Output
```

we have

```
Input

↓

Generate activation

↓

Apply activation

↓

Output
```

Now the neuron adapts to the input.

# Static vs Dynamic Neurons

## Static Neuron

```
Image

↓

Linear Layer

↓

ReLU

↓

Output
```

Every image uses

exactly

the same ReLU.

---

## Dynamic Neuron

```
Image

↓

Small Network

↓

Generate Parameters

↓

Dynamic ReLU

↓

Output
```

Every image gets its own activation function.

# Real World Analogy

Imagine a teacher.

A traditional teacher

using

exactly

the same method.

A good teacher changes

- speed
- examples
- explanation
- difficulty

depending on the student.

Dynamic ReLU behaves like the second teacher.

Every input gets

its own customized activation.

# Mathematical Review

A standard neuron computes

\[
z=Wx+b
\]

Then

\[
y=f(z)
\]

Normally

\[
f(z)=\max(0,z)
\]

Notice that

\(f\)

never changes.

It is identical for every sample.

# Dynamic ReLU Mathematics

Instead of

\[
f(x)=\max(0,x)
\]

Dynamic ReLU uses

\[
f(x)=\max(a_1x+b_1,\;a_2x+b_2)
\]

where

\[
a_1,a_2,b_1,b_2
\]

are **generated by another neural network**.

This means the activation function changes for every input.

# Where do these parameters come from?

Suppose

```
Feature Vector

↓

Small Neural Network

↓

[a₁,a₂,b₁,b₂]

↓

Dynamic ReLU
```

This small neural network is often called

- Parameter Generator
- Hyper Network
- Coefficient Generator

Its only job is to predict the best activation parameters for the current input.

# Why does this work?

Different images activate different neurons.

Examples

Cat

↓

Needs curved edge detectors.

---

Car

↓

Needs straight edge detectors.

---

Building

↓

Needs texture detectors.

Instead of forcing every image to use

the same activation,

Dynamic ReLU lets the network

adapt the activation

to the image.

# Summary

ReLU

✔ Fast

✔ Simple

❌ Fixed

---

PReLU

✔ Learnable

❌ Same activation for every image

---

Dynamic ReLU

✔ Learnable

✔ Different activation for every input

✔ Higher representation power

This idea forms the basis of many modern adaptive neural network architectures.

````python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("default")

x = np.linspace(-5, 5, 500)

plt.figure(figsize=(8,5))
plt.plot(x, np.maximum(0, x), linewidth=3)

plt.title("Standard ReLU")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)

plt.show()
````

````python
# Standard ReLU implementation

def relu(x):

    return np.maximum(0, x)


x = np.array([-3,-2,-1,0,1,2,3])

print(relu(x))
````

````python
# Dynamic ReLU (simplified)

def dynamic_relu(x, a, b):

    return np.maximum(a*x + b, 0)


x = np.linspace(-5,5,200)

y = dynamic_relu(x, 1.0, 0.0)

plt.figure(figsize=(8,5))

plt.plot(x,y)

plt.grid()

plt.title("Dynamic ReLU")

plt.show()
````

````python
# Different parameter choices

x=np.linspace(-5,5,500)

parameter_sets=[

    (1.0,0),

    (0.5,0),

    (2.0,0),

    (1.0,1),

    (1.0,-1)

]

plt.figure(figsize=(10,6))

for a,b in parameter_sets:

    plt.plot(

        x,

        dynamic_relu(x,a,b),

        label=f"a={a}, b={b}"

    )

plt.legend()

plt.grid()

plt.show()
````

````python
# Observe how changing 'a' changes slope

x=np.linspace(-4,4,400)

for a in [0.25,0.5,1,1.5,2]:

    plt.plot(

        x,

        dynamic_relu(x,a,0),

        label=f"a={a}"

    )

plt.legend()

plt.grid()

plt.title("Effect of slope")

plt.show()
````

````python
# Observe how changing bias shifts activation

x=np.linspace(-5,5,400)

for b in [-2,-1,0,1,2]:

    plt.plot(

        x,

        dynamic_relu(x,1,b),

        label=f"b={b}"

    )

plt.legend()

plt.grid()

plt.title("Effect of bias")

plt.show()
````

````python
# Two-piece Dynamic ReLU

def dynamic_relu_two_piece(x,a1,b1,a2,b2):

    y1=a1*x+b1

    y2=a2*x+b2

    return np.maximum(y1,y2)


x=np.linspace(-5,5,500)

y=dynamic_relu_two_piece(

    x,

    1,

    0,

    0.3,

    1

)

plt.figure(figsize=(8,5))

plt.plot(x,y)

plt.grid()

plt.title("Two-piece Dynamic ReLU")

plt.show()
````

````python
# Compare multiple Dynamic ReLU shapes

configs=[

    (1,0,0.5,0),

    (2,-1,0.5,1),

    (1,1,2,-2),

    (0.5,0,2,0)

]

x=np.linspace(-5,5,500)

plt.figure(figsize=(10,6))

for c in configs:

    plt.plot(

        x,

        dynamic_relu_two_piece(x,*c),

        label=str(c)

    )

plt.legend()

plt.grid()

plt.show()
````

````python
# Compare multiple Dynamic ReLU shapes

configs=[

    (1,0,0.5,0),

    (2,-1,0.5,1),

    (1,1,2,-2),

    (0.5,0,2,0)

]

x=np.linspace(-5,5,500)

plt.figure(figsize=(10,6))

for c in configs:

    plt.plot(

        x,

        dynamic_relu_two_piece(x,*c),

        label=str(c)

    )

plt.legend()

plt.grid()

plt.show()
````

````python
# Random Dynamic ReLU

np.random.seed(42)

x=np.linspace(-5,5,500)

plt.figure(figsize=(10,6))

for _ in range(8):

    a1=np.random.uniform(0,2)

    a2=np.random.uniform(0,2)

    b1=np.random.uniform(-2,2)

    b2=np.random.uniform(-2,2)

    plt.plot(

        x,

        dynamic_relu_two_piece(

            x,

            a1,

            b1,

            a2,

            b2

        )

    )

plt.title("Random Dynamic ReLU Shapes")

plt.grid()

plt.show()
````

````python
# ReLU vs Dynamic ReLU

x=np.linspace(-5,5,500)

plt.figure(figsize=(8,5))

plt.plot(

    x,

    relu(x),

    linewidth=3,

    label="ReLU"

)

plt.plot(

    x,

    dynamic_relu_two_piece(

        x,

        1,

        0,

        0.3,

        1

    ),

    linewidth=3,

    label="Dynamic ReLU"

)

plt.legend()

plt.grid()

plt.show()
````
